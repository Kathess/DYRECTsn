```python
import os
import pandas as pd
import numpy as np
from math import floor, ceil
```

```python
ncFolder = "dyrectsn_individualDelays_1GBit"
ncFiles = os.listdir(ncFolder)
print(ncFiles)
```

```python
# filter invalid files
minIndependDelay = dict()
maxIndependDelay = dict()
avgIndependDelay = dict()
minDependDelay = dict()
maxDependDelay = dict()
avgDependDelay = dict()
for file in ncFiles:
    with open(ncFolder + "/" + file, "r") as f:
        numDependDelays = 0
        numIndependDelays = 0
        minIndependDelay[file] = float("inf")
        maxIndependDelay[file] = 0
        avgIndependDelay[file] = 0
        minDependDelay[file] = float("inf")
        maxDependDelay[file] = 0
        avgDependDelay[file] = 0
        lines = f.readlines()
        # read lines in reverse
        for line in lines[::-1]:
            if "Successfully" in line:
                split = line.strip().split(":")[-1].split("of")
                if split[0].strip() != split[1].strip():
                    print(f"Error: reservation unsuccessful for {file}")
                    break
            if "Flow Delay" in line:
                delay = float(line.strip().split(":")[-1].strip().replace("ms",""))
                if "Dependent" in line:
                    minIndependDelay[file] = min(minIndependDelay[file], delay)
                    maxIndependDelay[file] = max(maxIndependDelay[file], delay)
                    avgIndependDelay[file] += delay
                    numIndependDelays += 1
                elif "Independent" in line:
                    minDependDelay[file] = min(minDependDelay[file], delay)
                    maxDependDelay[file] = max(maxDependDelay[file], delay)
                    avgDependDelay[file] += delay
                    numDependDelays += 1
        avgIndependDelay[file] /= numIndependDelays
        avgDependDelay[file] /= numDependDelays
# print(minIndependDelay)
print(maxIndependDelay)
# print(avgIndependDelay)
# print(minDependDelay)
# print(maxDependDelay)
# print(avgDependDelay)       
```

```python
minMaxIndependentDelay = float("inf")
minMaxFile = ""
for file in maxIndependDelay.keys():
    if maxIndependDelay[file] < minMaxIndependentDelay:
        minMaxIndependentDelay = maxIndependDelay[file]
        minMaxFile = file
print(minMaxIndependentDelay, minMaxFile)
```
